## Construct Offer and Demand of WITS in AI Model Input Matrix 

```python
import os
import pandas as pd

# File Definition 
WITS = 'aggregate_offer_demand_by_TP_WITS.csv'
WITS_read = pd.read_csv(WITS, delimiter=",")

# Rename the column "NZDTDateI" to "NZDTDate"
WITS_read.rename(columns={'NZDTDateI': 'NZDTDate'}, inplace=True)

# Convert "NZDTDate" column format to suitable one for mergency with other tables
# Convert the "NZDTDate" column to datetime format
WITS_read["NZDTDate"] = pd.to_datetime(WITS_read["NZDTDate"], format='%Y%m%d')
# Convert datetime format to string format with the desired format "YYYY-MM-DD"
WITS_read["NZDTDate"] = WITS_read["NZDTDate"].dt.strftime("%Y-%m-%d")

# Filter the DataFrame for BEN2201 and HAY2201 BusCodes
ben_df = WITS_read[WITS_read['BusCode'].str.startswith('BEN2201')]
hay_df = WITS_read[WITS_read['BusCode'].str.startswith('HAY2201')]

# Define a custom aggregation function
def custom_agg(x):
    if x.empty:
        return None
    else:
        return x.iloc[0]

# Define aggregation mappings for each band category and aggregate supply/demand
agg_mappings = {
    'Band_category': custom_agg,
    'Offer': 'sum',
    'AggregateDemand': 'first'
}

# Group by "NZDTDate" and "TradingPeriod" for BEN2201 and HAY2201
result_ben = ben_df.groupby(["NZDTDate", "TradingPeriod"]).agg(agg_mappings).add_suffix('_S')
result_hay = hay_df.groupby(["NZDTDate", "TradingPeriod"]).agg(agg_mappings).add_suffix('_N')

# Merge the results for South and North islands based on NZDTDate and TradingPeriod
result = pd.merge(result_ben, result_hay, on=["NZDTDate", "TradingPeriod"])

# Reset index to make "NZDTDate" and "TradingPeriod" as regular columns
result.reset_index(inplace=True)

# Create band columns for BEN2201
ben_bands = ben_df.pivot_table(index=["NZDTDate", "TradingPeriod"], columns="Band_category", values="Offer", aggfunc="sum").fillna(0)
ben_bands.columns = [f"Band{i}_S" for i in range(1, 6)]
result = pd.merge(result, ben_bands, on=["NZDTDate", "TradingPeriod"])

# Create band columns for HAY2201
hay_bands = hay_df.pivot_table(index=["NZDTDate", "TradingPeriod"], columns="Band_category", values="Offer", aggfunc="sum").fillna(0)
hay_bands.columns = [f"Band{i}_N" for i in range(1, 6)]
result = pd.merge(result, hay_bands, on=["NZDTDate", "TradingPeriod"])

# Drop unnecessary columns
result.drop(columns=['Band_category_S', 'Band_category_N'], inplace=True)

# Rearrange the columns and rename 'AggregateDemand_S' and 'AggregateDemand_N'
WITS_revised = result[['NZDTDate', 'TradingPeriod', 
                 'Band1_S', 'Band2_S', 'Band3_S', 'Band4_S', 'Band5_S',
                 'AggregateDemand_S',
                 'Band1_N', 'Band2_N', 'Band3_N', 'Band4_N', 'Band5_N',
                 'AggregateDemand_N']].rename(columns={'AggregateDemand_S': 'Demand_S', 'AggregateDemand_N': 'Demand_N'})

# # Rearrange the columns and rename 'AggregateDemand_S' and 'AggregateDemand_N'
# WITS_revised = result[['NZDTDate', 'TradingPeriod', 
#                  'Band1_S', 'Band2_S', 'Band3_S', 'Band4_S', 'Band5_S',
#                  'Offer_S', 'AggregateDemand_S',
#                  'Band1_N', 'Band2_N', 'Band3_N', 'Band4_N', 'Band5_N',
#                  'Offer_N', 'AggregateDemand_N']].rename(columns={'AggregateDemand_S': 'Demand_S', 'AggregateDemand_N': 'Demand_N'})


WITS_revised
```
